```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('raw_data.csv')
```

```python
data.head()
```

```python
data['Age'] = data['Age'].fillna(data['Age'].median())
```

```python
data['Score'] = data['Score'].replace('NA', np.nan).astype(float)
```

```python
data['Score'] = data['Score'].fillna(data['Score'].mean())
```

```python
data = data.dropna(subset=['Name', 'Category'])
```

```python
data = data.sort_values(by='ID', ascending=True)
```

```python
plt.figure(figsize=(18, 7))
sns.barplot(x='Name', y='Score', data=data)
plt.title('Scores by Individual')
plt.xticks(rotation=90)
plt.show()
```

```python
data.to_csv('cleaned_data.csv', index=False)
```
